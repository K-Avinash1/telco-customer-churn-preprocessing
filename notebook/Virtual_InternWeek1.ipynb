```python
categorical_columns = df.select_dtypes(include="object").columns

for column in categorical_columns:
    print(f"\n{column}:")
    print(df[column].unique())
```

```python
numerical_columns = ["tenure", "MonthlyCharges", "TotalCharges"]

for column in numerical_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)

    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_bound) |
        (df[column] > upper_bound)
    ]

    print(f"\nColumn: {column}")
    print(f"Q1: {Q1:.2f}")
    print(f"Q3: {Q3:.2f}")
    print(f"IQR: {IQR:.2f}")
    print(f"Lower Bound: {lower_bound:.2f}")
    print(f"Upper Bound: {upper_bound:.2f}")
    print(f"Number of outliers: {len(outliers)}")
```

```python
import matplotlib.pyplot as plt

for column in numerical_columns:
    plt.figure(figsize=(8, 4))
    plt.boxplot(df[column])
    plt.title(f"Box Plot of {column}")
    plt.ylabel(column)
    plt.show()
```

```python
df_cleaned = df.copy()
df_cleaned.shape
```

```python
for column in categorical_columns:
    print(f"\n{column}:")
    print(df_cleaned[column].unique())
```

```python
# Remove the customer ID because it is only an identifier
df_cleaned = df_cleaned.drop("customerID", axis=1)

# Convert binary categorical columns to numerical values
binary_columns = [
    "gender",
    "Partner",
    "Dependents",
    "PhoneService",
    "PaperlessBilling",
    "Churn"
]

for column in binary_columns:
    df_cleaned[column] = df_cleaned[column].map({
        "Yes": 1,
        "No": 0,
        "Male": 1,
        "Female": 0
    })
```

```python
df_cleaned[binary_columns].head()
```

```python
categorical_columns = [
    "MultipleLines",
    "InternetService",
    "OnlineSecurity",
    "OnlineBackup",
    "DeviceProtection",
    "TechSupport",
    "StreamingTV",
    "StreamingMovies",
    "Contract",
    "PaymentMethod"
]

df_cleaned = pd.get_dummies(
    df_cleaned,
    columns=categorical_columns,
    drop_first=True,
    dtype=int
)
```

```python
df_cleaned.head()
```

```python
print("Shape after encoding:", df_cleaned.shape)
```

```python
print(df_cleaned.dtypes)
```

```python
scaling_columns = [
    "tenure",
    "MonthlyCharges",
    "TotalCharges"
]
df_cleaned[scaling_columns].describe()
```

```python
from sklearn.preprocessing import MinMaxScaler

scaler = MinMaxScaler()

df_cleaned[scaling_columns] = scaler.fit_transform(
    df_cleaned[scaling_columns]
)
```

```python
df_cleaned[scaling_columns].head()
```

```python
df_cleaned[scaling_columns].describe()
```

```python
correlation_matrix = df_cleaned.corr()

churn_correlation = correlation_matrix["Churn"].sort_values(
    ascending=False
)

print(churn_correlation)
```

```python
import matplotlib.pyplot as plt

churn_correlation.drop("Churn").sort_values().plot(
    kind="barh",
    figsize=(10, 8)
)

plt.title("Feature Correlation with Churn")
plt.xlabel("Correlation Coefficient")
plt.ylabel("Features")
plt.show()
```

```python
feature_columns = [col for col in df_cleaned.columns if col != "Churn"]

duplicate_profiles = (
    df_cleaned[df_cleaned.duplicated(subset=feature_columns, keep=False)]
    .groupby(feature_columns)["Churn"]
    .nunique()
)

print("Duplicate feature profiles:", len(duplicate_profiles))
print(
    "Profiles with different Churn values:",
    (duplicate_profiles > 1).sum()
)
```
